### 1. Создание Series из списка

Создайте `Series` из списка чисел `[5, 15, 25, 35, 45]` с индексами `['a', 'b', 'c', 'd', 'e']`. Выведите его на экран и определите его тип данных.


````python
import pandas as pd

data = [5, 15, 25, 35, 45]
indices = ['a', 'b', 'c', 'd', 'e']

my_series = pd.Series(data=data, index=indices)

print("Созданный Series:")
print(my_series)
print("-" * 30)

print(f"Тип данных элементов внутри Series: {my_series.dtype}")
print(f"Тип самого объекта: {type(my_series)}")
````

### 2. Получение элемента Series

Дан `Series` с индексами `['A', 'B', 'C', 'D', 'E']` и значениями `[12, 24, 36, 48, 60]`. Используйте `.loc[]` для получения элемента с индексом `'C'` и `.iloc[]` для получения третьего элемента.


````python
import pandas as pd

indices = ['A', 'B', 'C', 'D', 'E']
values = [12, 24, 36, 48, 60]
my_series = pd.Series(data=values, index=indices)

element_loc = my_series.loc['C']
print(f"Элемент с индексом 'C' (.loc): {element_loc}")

element_iloc = my_series.iloc[2]
print(f"Третий элемент (.iloc): {element_iloc}")
````

### 3. Фильтрация данных с помощью логической индексации

Создайте `Series` из массива NumPy `np.array([4, 9, 16, 25, 36, 49, 64])`. Выберите только те элементы, которые больше 20, и выведите результат.


````python
import numpy as np
import pandas as pd

np_array = np.array([4, 9, 16, 25, 36, 49, 64])
my_series = pd.Series(np_array)

filtered_series = my_series[my_series > 20]

print("Элементы Series, которые больше 20:")
print(filtered_series)
````

### 6. Проверка пропущенных значений

Создайте `Series` с данными `[1.2, np.nan, 3.4, np.nan, 5.6, 6.8]`. Напишите код, который проверяет, есть ли в `Series` пропущенные значения (`NaN`), и выведите индексы таких элементов.


````python
import numpy as np
import pandas as pd

data = [1.2, np.nan, 3.4, np.nan, 5.6, 6.8]
my_series = pd.Series(data)

print(my_series.isna().any())

nan_indices = my_series[my_series.isna()].index
print(nan_indices)
````

### 8. Арифметические операции с Series

Создайте два `Series`:
* `s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])`
* `s2 = pd.Series([5, 15, 25, 35], index=['b', 'c', 'd', 'e'])`

Выполните сложение `s1 + s2`. Объясните, почему в результате появляются `NaN`, и замените их на `0`.


````python
import pandas as pd

s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s2 = pd.Series([5, 15, 25, 35], index=['b', 'c', 'd', 'e'])

result = s1 + s2
print(result)

result_filled = result.fillna(0)
print(result_filled)
````

### 12. Проверка уникальности индексов

Создайте `Series` с индексами `['A', 'B', 'A', 'C', 'D', 'B']` и значениями `[10, 20, 30, 40, 50, 60]`. Проверьте, являются ли индексы уникальными. Если нет, сгруппируйте повторяющиеся индексы и сложите их значения.


````python
import pandas as pd

indices = ['A', 'B', 'A', 'C', 'D', 'B']
values = [10, 20, 30, 40, 50, 60]

my_series = pd.Series(values, index=indices)

print(my_series.index.is_unique)

if not my_series.index.is_unique:
    result = my_series.groupby(level=0).sum()
    print(result)
````

### 13. Преобразование строковых дат в DatetimeIndex

Создайте `Series`, где индексами будут строки `['2024-03-10', '2024-03-11', '2024-03-12']`, а значениями — `[100, 200, 300]`. Преобразуйте индексы в `DatetimeIndex` и выведите тип данных индекса.

### 15. Построение графика на основе Series

Создайте `Series`, где индексами будут даты с 1 по 30 марта 2024 года, а значениями – случайные числа от 50 до 150. Постройте график значений с помощью `matplotlib`. Добавьте заголовок, подписи осей и сетку.


````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)

dates = pd.date_range(start="2024-03-01", end="2024-03-30", freq="D")
values = np.random.randint(50, 151, size=len(dates))
my_series = pd.Series(values, index=dates)

plt.figure(figsize=(10, 5))
plt.plot(my_series.index, my_series.values, marker="o")
plt.title("График случайных значений за март 2024")
plt.xlabel("Дата")
plt.ylabel("Значение")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

### Индивидуальное задание. Преобразование и анализ погодных данных

Создайте `Series`, где индексами будут даты с 10 по 20 марта 2024 года, а значениями – случайные температуры от -5 до +10 градусов. Выделите данные за 15-17 марта, преобразуйте индекс в `DatetimeIndex`, а затем постройте график температуры.


````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)

dates = pd.date_range(start="2024-03-10", end="2024-03-20", freq="D")
temperatures = np.random.randint(-5, 11, size=len(dates))
weather_series = pd.Series(temperatures, index=dates)

weather_series.index = pd.to_datetime(weather_series.index)

filtered_weather = weather_series["2024-03-15":"2024-03-17"]

plt.figure(figsize=(8, 4))
plt.plot(filtered_weather.index, filtered_weather.values, marker="o", color="orange")
plt.title("Изменение температуры с 15 по 17 марта 2024")
plt.xlabel("Дата")
plt.ylabel("Температура (°C)")
plt.grid(True)
plt.xticks(filtered_weather.index, filtered_weather.index.strftime("%Y-%m-%d"))
plt.tight_layout()
plt.show()
````
